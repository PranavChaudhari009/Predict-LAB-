```python
import pandas as pd
from textblob import TextBlob
from sklearn.model_selection import train_test_split
import re
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report, confusion_matrix,f1_score
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier
import pickle
```

```python
data = pd.read_csv("Twitter_Data.csv")
```

```python
data
```

```python
data.isnull().sum()
```

```python
data["category"].fillna(data["category"].mean(),inplace=True)
```

```python
data["clean_text"] = data["clean_text"].fillna("")
```

```python
data.isnull().sum()
```

```python
print(data.info())
print(data.describe())
print(data.duplicated().sum())
```

```python
x = data["clean_text"]
y = data["category"].round().astype(int) # Round to nearest integer and convert to int

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
#text preprocessing

#lowercasing

data["clean_text"] = data["clean_text"].str.lower()
data["clean_text"] = data["clean_text"].str.replace(r"\s+", " ", regex=True).str.strip()
data["clean_text"] = data["clean_text"].str.replace(r"http\S+|www\S+", "", regex=True)
data["clean_text"] = data["clean_text"].str.replace(r"[^\w\s]", "", regex=True)
```

```python
Vectorizer = TfidfVectorizer(   max_features=20000,
    ngram_range=(1,2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)
x_train_vect = Vectorizer.fit_transform(x_train) # Correctly fit and transform using TfidfVectorizer
x_test_vect  = Vectorizer.transform(x_test)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(x_train_vect,y_train)
y_pred = model.predict(x_test_vect)
```

```python
accuracy = accuracy_score(y_pred,y_test)
print("accuracy_Score:", accuracy)
```

```python
print("classification report:",classification_report(y_pred,y_test))
print("confusion matrix:",confusion_matrix(y_pred,y_test))
print("Marco_score:",f1_score(y_pred,y_test, average="macro"))
```

```python
#second model
model = MultinomialNB()

model.fit(x_train_vect,y_train)
y_pred = model.predict(x_test_vect)

print("accuracy score:",accuracy_score(y_pred,y_test))
```

```python
model = RandomForestClassifier()

model.fit(x_train_vect,y_train)
y_pred = model.predict(x_test_vect)

print("accuracy score:",accuracy_score(y_pred,y_test))
```
